```python
import pandas as pd
```

```python
sales = pd.read_csv('/content/supermarket_sales (1).csv')

sales.head()
```

Data Cleaning

```python
sales.info()
```

```python
#Checking for missing values

print(sales.isnull().sum())

#Replacing null values with median

sales['Rating'].fillna(sales['Rating'].median(), inplace=True)
```

```python
#Checking for Duplicates

print(sales.duplicated().sum())
```

```python
#Standardizing column headers

sales.columns = sales.columns.str.lower()
```

```python
#Correctly Formatting the table

sales['date'] = pd.to_datetime(sales['date'])
sales[['total', 'cogs', 'tax 5%', 'gross income']] = sales[['total', 'cogs', 'tax 5%', 'gross income']].astype(float)

sales.info()
```

```python
# Fixing inconsistent city entries
branch_to_city = {
    'A': 'Naypidaw',
    'B': 'Mandalay',
    'C': 'Yangon'
}


sales['city'] = sales['branch'].map(branch_to_city)

print(sales[['branch', 'city']].drop_duplicates())
```

```python
# Making sure all rating values are between the range of 0-10
sales.loc[sales['rating'] > 10, 'rating'] = sales['rating'] / 10

sales = sales[(sales['rating'] >= 1) & (sales['rating'] <= 10)]

# Check the cleaned d
sales.info()
```

```python
#Seperating the date column for further analysis

sales['day'] = sales['date'].dt.day
sales['month'] = sales['date'].dt.month
sales['year'] = sales['date'].dt.year

sales.info()
```

```python
#Standardizing Categorical Data

sales['branch'] = sales['branch'].str.strip().str.upper()
sales['city'] = sales['city'].str.strip().str.title()
sales['payment'] = sales['payment'].str.strip().str.capitalize()
```

```python
#Saving the file

sales.to_csv("cleaned_retail_sales.csv", index=False)
```
